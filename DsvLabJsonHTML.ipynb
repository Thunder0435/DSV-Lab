```python
import pandas as pd
import numpy as np

# Creating sample data with missing values
data = {
    "Name": ["Aarav", "Riya", "Kabir"],
    "Maths": [85, np.nan, 92],
    "Science": [90, 78, 88]
}

df = pd.DataFrame(data)

# Writing the file using ; as the delimiter
# Replacing missing values with -999
df.to_csv(
    "marks.csv",
    sep=";",
    index=False,
    na_rep="-999"
)

# Read the file and convert -999 into NaN
df_loaded = pd.read_csv(
    "marks.csv",
    sep=";",
    na_values="-999"
)

print(df_loaded)
```

```python
# Store results from each chunk
partial_results = []

# Read the file in chunks of 2000 rows
for chunk in pd.read_csv("taxis.csv", chunksize=2000):

    # Calculate sum and count for each payment type
    stats = chunk.groupby("payment")["tip"].agg(["sum", "count"])

    # Store the results
    partial_results.append(stats)

# Combine results from all chunks
totals = pd.concat(partial_results).groupby(level=0).sum()

# Calculate the overall mean
totals["mean_tip"] = totals["sum"] / totals["count"]

# Display the final result
result = totals[["mean_tip"]].reset_index()

print(result)
```

```python
import pandas as pd

# Create book data
data = {
    "Book": ["Python Basics", "Data Science", "Machine Learning"],
    "Pages": [250, 320, 400],
    "Price": [450, 600, 800]
}

df = pd.DataFrame(data)

# Export DataFrame to JSON
df.to_json("books.json", orient="records", indent=2)

# Read the JSON file
df1 = pd.read_json("books.json", orient="records")

# Check whether both DataFrames are equal
print(df.equals(df1))
```
